# Muat dataset ke HDFS

### Masuk ke namenode :

docker exec -it namenode bash

### Buat folder untuk raw data :

hdfs dfs -mkdir -p /data/raw/fintech

### Pindahkan dataset ke container namenode :

docker cp fraudTrain.csv namenode:/tmp/

### Upload dataset ke HDFS:


hdfs dfs -put /tmp/fraud.csv /data/raw/fintech/

### Cek data di folder :

hdfs dfs -ls /data/raw/fintech

# Import

```python
from pyspark.sql import SparkSession
```

# Konfigurasi

```python
spark = SparkSession.builder \
    .appName("Ingestion") \
    .master("local[*]") \
    .config("spark.driver.memory", "2g") \
    .config("spark.executor.memory", "2g") \
    .config("spark.sql.shuffle.partitions", "8") \
    .getOrCreate()

spark
```

# Output

```python
df = spark.read.csv("hdfs://namenode:9000/data/raw/fintech/fraudTrain.csv",header=True,inferSchema=True)
df.show(5)
```
